```python
# Sales/Profit missing values -> kept as NaN intentionally (no fill, no drop)
# pandas aggregate functions (.sum(), .mean()) automatically skip NaN, same as SQL NULL behavior
```

```python
# Data Standardization
```

```python
# # Remove "~?" invalid characters from Sales
data["Sales"] = data["Sales"].str.replace("~?", "", regex=False)
```

```python
# Remove extra spaces
data["Ship Mode"] = data["Ship Mode"].str.strip()
```

```python
# Remove extra spaces from Customer Name
data["Customer Name"] = data["Customer Name"].str.strip()
```

```python
# Standardize Text Values 
data["Region"] = data["Region"].str.title()
```

```python
# Correct spelling mistakes
data['Sub-Category'] = data['Sub-Category'].replace({
    'Furnishingss': 'Furnishings',
    'Bookcasess': 'Bookcases',
    'Storagee': 'Storage',
    'Appliancess': 'Appliances',
    'Machiness': 'Machines',
    'Accessoriess': 'Accessories'
})
```

```python
# Standardize City names (Remove State name from City) - generic fix

data["City"] = data["City"].str.split(",").str[0]
```

```python
# Replace invalid text
data["Profit"] = data["Profit"].replace(["unknown", "N/A"], pd.NA)
```

```python
# Duplicate Value
```

```python
print(data.duplicated())
```

```python
data.duplicated().sum()
```

```python
# Data Type Conversion
```

```python
data["Order Date"]=pd.to_datetime(data["Order Date"],format="mixed", dayfirst=True,errors="coerce")
data["Ship Date"]=pd.to_datetime(data["Ship Date"],format="mixed", dayfirst=True,errors="coerce")
data["Sales"] = pd.to_numeric(data["Sales"], errors="coerce")
data["Profit"] = pd.to_numeric(data["Profit"], errors="coerce")
```

```python
# Invalid Value correction
```

```python
# Negative Sales - Verify first
data[data['Sales'] < 0]
```

```python
# Drop rows with negative Sales 
data = data[data['Sales'] >= 0]
```

```python
# Discount > 1 - Verify first
data[data['Discount'] > 1]
```

```python
# Invalid Discount (>1 or <0) -> null only this column, keep the row
data.loc[data['Discount'] > 1, 'Discount'] = np.nan
data.loc[data['Discount'] < 0, 'Discount'] = np.nan
```

```python
# Negative Quantity - Verify first
data[data['Quantity'] < 0]
```

```python
# Fix negative Quantity (sign-entry inconsistency)
data['Quantity'] = data['Quantity'].abs()
```

```python
# Duplicate Removal (after all cleaning is done)
```

```python
data = data.drop_duplicates(subset=data.columns.drop(["Row ID", "Order ID"]))
```

```python
# Line Plot - Monthly Sales Trend
data.groupby(data['Order Date'].dt.to_period("M"))['Sales'].sum().plot(kind='line')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

# Insights - line  plot

# Pattern
- Sales go up and down every month from 2019 to 2022, no steady increase or decrease.
- Some months (like early 2021) show very high sales, while some months drop a lot.
# Reason
- The chart only shows monthly sales numbers, it doesn't explain what caused these ups and downs.
# Suggestion
- Look at sales by Category, Sub-Category, Region, or Segment to understand why certain months performed better or worse.

```python
# Bar Plot - Category-wise Sales
data.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.title('Category-wise Sales')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

# Insights - Bar plot

# Pattern
- Office Supplies has the highest sales, while Furniture is almost equal (less than 0.3% difference).
- Technology has comparatively lower sales (around 3–3.5% lower than the other two categories).
# Reason
- The chart compares total sales by category only. It does not explain why Technology sales are lower.
# Suggestion
- Analyze Sub-Category and Product Name level sales to identify which products contribute more or less to Technology sales.

```python
# Scatter Plot - Discount vs Profit
plt.scatter(data['Discount'], data['Profit'])
plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.show()
```

# Insights - Scatter plot

# Pattern
- Discount only occurs at fixed levels — 0.0, 0.1, 0.2, 0.3, and 0.5 (no orders at 0.4).
- As discount increases, the maximum profit value decreases — at 0 discount, profit goes up to ~6800, while at 0.5 discount, it only reaches ~2600.

# Reason
- The chart suggests a negative relationship between Discount and Profit, but it does not confirm the exact reason

# Suggestion
- Analyze high-discount orders (0.3 and 0.5) separately at the Category, Sub-Category, and Product Name level to identify which discounts are causing losses.

```python
# Box Plot - Profit by Category
sns.boxplot(x='Category', y='Profit', data=data)
plt.title('Profit by Category')
plt.show()
```

# Pattern

- Technology has some of the highest positive profit outliers.
- All three categories contain many outliers (both high-profit and high-loss orders).

# Reason
- The chart shows that profit varies across all categories.
- The chart alone cannot explain why these outliers occur .

# Suggestion
- Analyze the outlier orders separately to identify the reason for very high profits and large losses.
- Compare Product Name and Sub-Categories within each category for deeper analysis.

```python
# Histplot - Sales Distribution
sns.histplot(data['Sales'])
plt.title('Sales Distribution')
plt.show()
```

# Insights - Histplot

# Pattern
- Most sales transactions are concentrated at lower sales values.
- As the sales amount increases, the number of transactions gradually decreases.
- A small number of transactions have very high sales values, forming a long right tail.
- The sales distribution is positively (right) skewed.

# Reason
- The histogram only shows how sales values are distributed.
- It does not explain why some transactions have much higher sales than others.

# Suggestion
- Analyze the high-value sales transactions separately to identify which Product Name, Category, or Customer Name contribute to them.
- Compare low-value and high-value transactions to understand the business pattern.

# Key Findings
- Multiple data-quality issues were identified and corrected, including Region inconsistencies, Sub-Category typos, invalid Discount values, and negative Sales/Quantity values.
- Monthly Sales fluctuated across 2019–2022 with no consistent upward or downward trend.
- Office Supplies and Furniture recorded nearly equal Sales, while Technology generated slightly lower Sales.
- Higher discount levels showed lower maximum Profit values, although the overall Discount–Profit relationship remained weak.
- Sales and Quantity showed a moderate positive relationship.
- All categories contained both high-profit and high-loss outliers, with Technology showing some of the highest positive profit outliers.
- Most Sales transactions were low-value, with a few high-value orders forming a long right tail.

# Business Recommendations
- Review high-discount orders (especially 0.3 and 0.5) to understand their impact on profitability.
- Investigate high-profit and high-loss outlier orders to understand why they performed differently.
- Analyze high-value Sales orders to understand which categories or regions generate the most Sales.
- Check Category and Sub-Category performance before making business decisions.
